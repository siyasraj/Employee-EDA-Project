```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Age",bins=10, kde=True)
plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()
```

```python
##The chart shows the age distribution of employees. Most employees are between the younger and middle age groups. There are fewer employees at the very high age range. This helps HR understand the age structure of employees.
```

```python
sns.countplot(x='Department', data=df)
plt.title('Number of Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```

```python
##The chart shows the number of employees in each department. IT and Sales have more employees compared with some other departments. HR has fewer employees in this dataset. This helps HR understand the workforce distribution.
```

```python
sns.countplot(x='Gender', data=df)
plt.title('Gender Distribution of Employees')
plt.xlabel('Gender')
plt.ylabel('Number of Employees')
plt.show()
```

```python
##The chart shows the number of employees by gender. Male and female employees make up most of the workforce. The number of employees in the Other category is much smaller. This helps HR understand the gender distribution.
```

```python
sns.histplot(df['AnnualSalary'], bins=20, kde=True)
plt.title('Annual Salary Distribution')
plt.xlabel('Annual Salary')
plt.ylabel('Number of Employees')
plt.show()
```

```python
##The chart shows the distribution of annual salaries. Most employees have salaries in the lower range. A few employees have very high salary values. This shows that salary values vary widely among employees.
```

```python

sns.scatterplot(x='ExperienceYears', y='AnnualSalary', data=df)
plt.title('Experience vs Annual Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Annual Salary')
plt.show()
```

```python
##The chart compares employee experience with annual salary. Most employees are grouped in the lower salary range. Some employees have unusually high salary values. This helps HR study the relationship between experience and salary
```

```python
plt.figure(figsize=(10,6))
plt.bar(data=df,x="ProjectsCompleted",height="WeeklyHours")
plt.title("WeeklyHours Distribution")
plt.xlabel("W")
plt.ylabel("Number of employees")
plt.show()
```

```python
##The chart shows the distribution of employees based on weekly working hours. Most employees have weekly hours around the normal working range. The number of employees changes across different working-hour values. This helps HR understand employee workload.
```

```python
df[df["ProjectsCompleted"]==1]["WeeklyHours"].plot(kind="hist")
```

```python
##The chart shows weekly working hours for employees who completed 1 project. Most of these employees have working hours around the middle range. Very few employees are at the extreme ends. This helps HR understand the workload of this group.
```

```python
df[df["ProjectsCompleted"]==5]["WeeklyHours"].plot(kind="hist")
```

```python
##The chart shows weekly working hours for employees who completed 5 projects. Most employees are concentrated around the middle working-hour range. Only a small number of employees have very low or very high working hours. This helps HR compare workload among employees.
```
